```python
# Question 9: Write a Python program to generate 1000 random numbers from a normal
# distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
# deviation using NumPy, and draw a histogram to visualize the distribution.

# importing numpy and matplotlib
import numpy as np
import matplotlib.pyplot as plt

# Generating random data
data = np.random.normal(50, 5, 1000)

# Print mean and standard deviation
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))

# Plot histogram
plt.hist(data, bins=30, color='purple')
plt.title("Normal Distribution (mean=50, std=5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# Question 10: You are working as a data analyst for a retail company. The company has
# collected daily sales data for 2 years and wants you to identify the overall sales trend.
# daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
#  235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
# ● Explain how you would apply the Central Limit Theorem to estimate the average sales
# with a 95% confidence interval.
# ● Write the Python code to compute the mean sales and its confidence interval.

"""First calculate the mean of sales, then measure how much the sales vary using standard deviation, and then
use a formula to create a confidence interval """

# import statistics and math
import statistics
import math

# given data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# mean
mean_sales = statistics.mean(daily_sales)

# standard deviation
std_dev = statistics.stdev(daily_sales)

# size of data
n = len(daily_sales)

# Z-score for 95% confidence
z = 1.96

# Calculate margin of error
margin_error = z * (std_dev / math.sqrt(n))

# Confidence interval
lower = mean_sales - margin_error
upper = mean_sales + margin_error

# Print results
print("Average daily sales:", mean_sales)
print(f"95% Confidence Interval ({lower},{upper})")
```
