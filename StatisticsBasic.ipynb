Question 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.


---
Answer:Descriptive statistics organize analyze and summarize data in such a way that describes the data in some manner. Inferential statistics makes inferences about the population using data drawn from population
- Descriptive statistics explains already known data and limited to a sample having a small size population, while inferenctiall statistics attempts to reach the conclusion about the population.
- Descriptive statistics used for describing trends while inferential statistics used for predicting trends.
- Descriptive statistics include mean, median, mode, variance, standard deviation, quartiles , etc.
- Inferential statistics include hypothesis testing, regression models, probability measures, etc.
Descriptive example:- calculating average height of 12 players. Inferential example:- random sample of 500 players from across the state and calculating their average height to estimate true average height of all players in the state.

Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.


---
Answer: Sampling in statistics involves selecting a subset of individuals from a population to estimate characteristics of the whole group.
- Random Sampling - every individual in the population has an equal chance of being selected. Selection is done entirely by chance like generating random numbers.
- Stratified Sampling - population is divided into two distinct subgroups(strata) based on shared characteristics and sample are drawn from each groups or stratas randomly.Selection can be propotional or equal.



Question 3: Define mean, median, and mode. Explain why these measures of central
tendency are important.


---
Answer: Mean , Median and Mode are measures of central tendency comes under descriptive statistics type.
- Mean:- it is calculated by summing all the values in a dataset and dividing by the total number of values. It'sensitive to outliers and is commonly used when the data is symmetrically distributed.
- Median:- The median is the middle value when the dataset is arranged in ascending or descending order. If there's an even number of values, it's the average of the two middle values. The median is robust to outliers and is often used when the data is skewed.
- Mode:- The mode is the value that occurs most frequently in the dataset. Unlike the mean and median, the mode can be applied to both numerical and categorical data. It's useful for identifying the most common value in a dataset.
Why these are important?:-
- Summarize the data
- Compare the distributions
- Gain Insights from data

Question 4: Explain skewness and kurtosis. What does a positive skew imply about the data?


---
Answer:Skewness measures symmetricity in data,kurtosis measures the tailedness.
- Skewness - it shows how much a distribution is far from symmetricity.If distribution is symmetric, it is zero skewed. A positive skew (also called right skew) means the tail on the right side is longer.This often results in the mean being greater than the median. A negative skew (also called left skew) means tail on the left side is longer.This results in the mean is less than the median.
- kurtosis measures the tailedness or the propensity of a distribution to produce outliers. Distributions with high kurtosis have heavy tails and more extreme outliers, while those with low kurtosis have light tails and fewer outliers.
- A Positive skew implies that in distribution plot tail is longer on right side which often indicated about result that mean is greater then median or mean>=median>=mode.



```python
# Question 5: Implement a Python program to compute the mean, median, and mode of a given list of numbers.
# numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
# Answer:

# import statistics
import statistics
# using given dataset
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
# print mean,median,mode
print("Mean =", statistics.mean(numbers))
print("Median =", statistics.median(numbers))
print("Mode =", statistics.mode(numbers))
```

```python
# Question 6: Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:
# list_x = [10, 20, 30, 40, 50]
# list_y = [15, 25, 35, 45, 60]
# Answer:

# import pandas
import pandas as pd
# using given list
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
# coverting list to dataframe
df = pd.DataFrame({'x':list_x,'y':list_y})
# calculating covariance and correlation
cov_xy = df['x'].cov(df['y'])
corr_xy = df['x'].corr(df['y'])
# print covariance and correlation
print("covariance=",cov_xy)
print("correlation=",corr_xy)
```

```python
# Question 7: Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:
# data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
# Answer:

# import matplotlib,seaborn,numpy
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
# using given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
# creating a box plot
plt.figure()
sns.boxplot(data,color="skyblue")
plt.title("Box Plot")
plt.xlabel("Values")
plt.ylabel("data")
plt.show()
# Calculate Inter Quartile Range
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1
# calculating lower fence and upper fence
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
# predicting Outliers that are below range of lower fence and upper range of upper fence
outliers = [x for x in data if x < lower or x > upper]
# printing the values
print("Q1:", q1)
print("Q3:", q3)
print("Inter Quartile Range:", iqr)
print("Lower Fence:", lower)
print("Upper Fence:", upper)
print("Outliers:", outliers)
```

```python
# Question 8: You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and  daily sales.
# ● Explain how you would use covariance and correlation to explore this relationship.
# ● Write Python code to compute the correlation between the two lists:
# advertising_spend = [200, 250, 300, 400, 500]
# daily_sales = [2200, 2450, 2750, 3200, 4000]
# Answer:
"""To explore relationship between advertising spend and daily sales first i will calculate the correlation and covariance then predict thing according to result.
Covariance provides a measure of how the two variables change together, if the value is positive, it suggests that as advertising spend increases, daily sales tend to increase as well.
However, covariance alone doesn't indicate the strength of this relationship. That’s where the correlation helps.Correlation gives result netween -1 to 1. If it's close to 1 it means
more advertising and more sales. if it's close to -1 means more advertising and fewer sales. If by chance it is zero then it shows no relationship.  """
# import pandas,seaborn,matplotlib
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# using given dataset
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
# making data frame out of given data
df = pd.DataFrame({"Advertising Spend":advertising_spend,"Daily Sales":daily_sales})
Correlation = df['Advertising Spend'].corr(df['Daily Sales'])
# creating regression plot
plt.figure()
sns.regplot(x = 'Advertising Spend', y = 'Daily Sales', data = df, color = 'Purple' , marker= 'o')
plt.text(250, 3800, f'Correlation: {Correlation:.2f}', fontsize=12, color='darkred')
plt.title("Advertising Spend vs Daily Sales")
plt.xlabel('Advertising Spend')
plt.ylabel('Daily Sales')
plt.show()

"""INTERPRETATION: The scatter plot with a regression line shows a clear upward trend,as advertising spend increases, daily sales also rise.
The correlation coefficient is 0.99, which means the relationship is very strong and positive."""
```
