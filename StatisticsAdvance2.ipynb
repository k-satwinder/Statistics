Question 1: What is hypothesis testing in statistics?


---
Answer: Hypothesis testing is a statistical method used to make decisions or inferences about population parameters based on sample data. It helps determine whether a claim or assumption about a population is likely to be true.


Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?


---
Answer: The null hypothesis denoted as H0 represents a default assumption about a population parameter. It typically states that there is no effect, no difference, or no relationship between variables.Alteernative hypothesis is opposite of null hypothesis and denoted as HA.



Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.


---
Answer: The significance level in hypothesis testing represents the probability of rejecting the null hypothesis when it is actually true.It is the threshold for deciding whether the observed data is statistically unusual enough to give a conclusion.If pvalue is less than equal to significance level then conclusion is to reject the null hypothesis else the conclusion is failed to reject the null hypothesis.




Question 4: What are Type I and Type II errors? Give examples of each.


---
Answer:
- A Type I error occurs when you reject the null hypothesis when it is actually true.
- A Type II error happens when you fail to reject the null hypothesis when it is actually false.
- Example : The average delivery time of products is 30 minutes or less.
- Type I error : conclusion was , delivery time is more than 30 minutes, and reject null hypothesis but in reality, the average is still 30 minutes or less.
- Type II error: conclusion was delivery time is 30 minutes or less, and fail to reject null hypothesis but in reality, the average is more than 30 minutes.






Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.


---
Answer:
- A Z-test is used to determine whether there is a significant difference between a sample mean and a population mean when the population standard deviation is known and the sample size is large typically n > 30. It uses the standard normal distribution (Z-distribution) to calculate the probability of observing the sample result under the null hypothesis.
- A T-test is used to compare a sample mean to a population mean or compare two sample means when the population standard deviation is unknown and the sample size is small typically n ≤ 30. It uses the t-distribution, which accounts for more variability in smaller samples by having heavier tails than the normal distribution.

```python
# Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in
# Python. Show the Python code and interpret the results.
# sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
# 50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
# 50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
# 50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

import numpy as np
from scipy.stats import norm

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,50.3, 50.4, 50.0, 49.7, 50.5, 49.9]


mu = 50
sigma = 0.5
n = len(sample_data)
sample_mean = np.mean(sample_data)
z = (sample_mean - mu) / (sigma / np.sqrt(n))
p_value = 2 * (1 - norm.cdf(abs(z)))

print(f"Z-statistic = {z}")
print(f"P-value = {p_value}")

if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
# Question 8: Write a Python script to simulate data from a normal distribution and
# calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(100, 15, 100)

mean = np.mean(data)
std = np.std(data)
lower = mean - 1.96 * (std / np.sqrt(len(data)))
upper = mean + 1.96 * (std / np.sqrt(len(data)))

print("Mean:", round(mean, 2))
print("95% Confidence Interval:", round(lower, 2), "to", round(upper, 2))

plt.hist(data, bins=10, color='skyblue')
plt.title("Normal Data with 95% Confidence Interval")
plt.show()
```

```python
# Question 9: Write a Python function to calculate the Z-scores from a dataset and
# visualize the standardized data using a histogram. Explain what the Z-scores represent
# in terms of standard deviations from the mean.

from statsmodels.stats.weightstats import ztest
import numpy as np
import matplotlib.pyplot as plt

null_mean = 0.09
population = np.random.randint(10,20,50)
zscore, pvalue = ztest(population, value=null_mean, alternative='larger')

print("Z-score:", zscore)

plt.hist(population, bins=20 ,color='r')
plt.title("Histogram")
plt.show()
```
